```python
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer, pipeline
from datasets import load_dataset
```

```python
# Load the GPT-2 model and tokenizer
model_name = "gpt2"
model = AutoModelForCausalLM.from_pretrained(model_name)
tokenizer = AutoTokenizer.from_pretrained(model_name)

# Sample text
text = "The quick brown fox jumps over the lazy dog."
inputs = tokenizer(text, return_tensors="pt")
outputs = model(**inputs, labels=inputs["input_ids"])

# Calculate perplexity
loss = outputs.loss
perplexity = torch.exp(loss)
print(f"Perplexity of the text: {perplexity.item()}")
```

```python
# Download the IMDB dataset
dataset = load_dataset("imdb", split="test[:1000]")
print("Dataset downloaded and loaded successfully!")
```

```python
# Load a pre-trained sentiment analysis pipeline
classifier = pipeline("sentiment-analysis")

# Map numeric labels to string labels
label_map = {0: "NEGATIVE", 1: "POSITIVE"}

# Maximum token length for the model
max_length = 512

# Evaluate accuracy
correct = 0
print("Evaluating sentiment analysis accuracy...")
for example in dataset:
    # Truncate text if it's too long
    truncated_text = example["text"][:max_length]
    prediction = classifier(truncated_text)[0]["label"]
    true_label = label_map[example["label"]]
    if prediction.upper() == true_label.upper():
        correct += 1

accuracy = correct / len(dataset)
print(f"Task Accuracy: {accuracy * 100:.2f}%")
```
